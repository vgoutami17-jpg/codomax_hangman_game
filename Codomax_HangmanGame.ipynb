```python
import random
```

```python
words = [
    "python",
    "computer",
    "programming",
    "keyboard",
    "internet"
]
```

```python
word = random.choice(words)

print("Welcome to Hangman!")
print("Guess the word one letter at a time.")
```

```python
guessed_word = ["_"] * len(word)
incorrect_guesses = 0
guessed_letters = []

max_incorrect_guesses = 6
```

```python
while incorrect_guesses < max_incorrect_guesses and "_" in guessed_word:

    print("\nWord:", " ".join(guessed_word))
    print("Incorrect guesses:", incorrect_guesses, "/", max_incorrect_guesses)

    guess = input("Guess a letter: ").lower()

    if len(guess) != 1 or not guess.isalpha():
        print("Please enter only one letter.")
        continue

    if guess in guessed_letters:
        print("You already guessed that letter.")
        continue

    guessed_letters.append(guess)

    if guess in word:
        print("Correct guess!")

        for i in range(len(word)):
            if word[i] == guess:
                guessed_word[i] = guess

    else:
        incorrect_guesses += 1
        print("Incorrect guess!")
```

```python
print("\n================================")

if "_" not in guessed_word:
    print("Congratulations! You guessed the word!")
    print("The word was:", word)
else:
    print("Game Over!")
    print("You reached 6 incorrect guesses.")
    print("The word was:", word)

print("================================")
```

## Project Description

The Hangman Game is a simple text-based Python game developed as part of the Codomax Python Programming Internship.

The player has to guess a randomly selected word one letter at a time.

The game provides a maximum of 6 incorrect guesses. The player wins by correctly guessing all the letters of the selected word.

## Features

- Randomly selects a word
- Uses 5 predefined words
- Allows the player to guess one letter at a time
- Allows a maximum of 6 incorrect guesses
- Displays the progress of the hidden word
- Prevents repeated letter guesses
- Displays a winning message
- Displays a game-over message

- Random module
- Lists
- Strings
- While loop
- For loop
- If-else statements
- User input
